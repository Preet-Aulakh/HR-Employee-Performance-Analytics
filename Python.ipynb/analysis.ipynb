```python
df_grouped =df.groupby("Department")["PerformanceRating"].mean().reset_index()
df_grouped.plot(kind="bar", x="Department", y="PerformanceRating")
plt.show()
```

```python
df_grouped=df.groupby("City")["AnnualSalary_INR"].mean().reset_index()
df_grouped.plot(kind="barh" , x="City", y="AnnualSalary_INR")
plt.show()
```
